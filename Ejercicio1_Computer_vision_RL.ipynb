```python
import numpy as np
import cv2
from google.colab.patches import cv2_imshow

```

```python
alto, ancho = 400, 400
```

```python
imagen = np.zeros((alto, ancho, 3), np.uint8)
```

```python
color_trazo = (255, 255, 0)
grosor = 15
```

```python
cv2.line(imagen, (120, 100), (120, 300), color_trazo, grosor)
```

```python
cv2.ellipse(imagen, (120, 150), (60, 50), 0, -90, 90, color_trazo, grosor)
```

```python
cv2.line(imagen, (140, 200), (200, 300), color_trazo, grosor)
```

```python
cv2_imshow(imagen)
```
