```python
import numpy as np
import cv2
from google.colab.patches import cv2_imshow
from matplotlib import pyplot as plt
```

```python
# Definimos tamaño de la imagen
alto, ancho = 400, 400

#Definimos la imagen
# 3 es el espacio de color RGB
imagen = np.zeros((alto, ancho, 3), np.uint8)

# Azul = 255 , Verde = 255, Rojo = 255
imagen[50,75] =  [255,255,255]

print(imagen[50,75])
```

```python
imagen_gris = np.zeros((100, 200,3), np.uint8)
imagen_gris[25:75, 50:150] = [255,255,0]

valor_pixel =  imagen_gris[50, 100]
print(valor_pixel)
# Mostramos la imagen
cv2_imshow(imagen_gris)
```

## Modificar una imagen

```python
# Leer imagen
imagen = cv2.imread("diablada.jpg")

# Convertimos a escala de gris
for y in range(imagen.shape[0]):
  for x in range(imagen.shape[1]):
    azul, verde, rojo = imagen[y,x]
    gris = int(azul * 0.114 + verde * 0.587 + rojo * 0.299)
    imagen[y,x] = [gris, gris, gris]

# Mostramos la imagen
cv2_imshow(imagen)
```

```python
# Observamos el histograma de colores de la imagen
imagen = cv2.imread("diablada.jpg")
imagen_gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

histograma = cv2.calcHist([imagen_gris], [0], None, [256], [0,256])
plt.plot(histograma)
plt.show()
```
